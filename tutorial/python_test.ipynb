```python
dummy_data = np.random.rand(200, 20)
```

If you run the next cell, you should see an animated plot of random connected points!

```python
plt.ion()

fig, ax = plt.subplots()
ax.set(xlim=(0,20), ylim=(0,1), title=f"Step {0:4.0f}")
line, = ax.plot(dummy_data[0], "o-", c='k')

for i in range(1,dummy_data.shape[0]):
    line.set_ydata(dummy_data[i])
    ax.set_title(f"Step {i:4.0f}")
    clear_output(wait=True)
    display(fig)
```
