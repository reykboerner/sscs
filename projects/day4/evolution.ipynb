The source code for this exercise is contained in the files `evolution_abm.py` and `abm_utils.py`, which can be loaded using the following commands.

This provides the classes `AgentEvolutionModel` and `ABM_Parameters`, as well as convenience functions for data analysis.

```python
from evolution_abm import AgentEvolutionModel, ABM_Parameters
from abm_utils import get_agent_positions, get_agent_properties
```

The `ABM_Parameters` class allows us to specify the parameter settings of the agent-based model (`AgentEvolutionModel`). You can either use the default values, `ABM_Parameters()`, or change the parameters by passing them as keyword arguments. For example, if you want to change the number of initial agents, write

`par = ABM_Parameters(N_init=200)`.

```python
par = ABM_Parameters()
par.show()
```

As a user, you also need to define the food input into the domain at each time step. This is done with the `food_input` function, which takes two positional arguments: the time step `t` and the number of agents `n`. This function outputs the number of food items, which are then distributed randomly across the domain.

For now, the function below simply returns 10 food items at each time step, so the food input depends neither on time nor the number of agents. However, you are free to change this - just make sure the modified function has exactly the same call signature and returns an integer number representing the food quantity.

```python
def food_input(t, n):
    """
    Specifies the food input (number of food items) into the domain
    at each time step. Can be a function of time `t` and of the
    population size `N`.
    """
    return 10
```

The agent-based model is initialized based on the parameters `par` and input function `food_input`.

```python
model = AgentEvolutionModel(par, food_input)
```

To run a simulation, use the `simulate` function as below. If you simulate many timesteps, live plotting can slow things down considerably. You can either turn live plotting off (`live_plot=False`) or control the plotting frequency (`plot_interval=...`).

The `simulate` function returns three objects, here stored as variables `agents`, `positions` and `food`.
- `agents` is a list of `len(timesteps+1)`, where each list element is itself a list of `Agent` objects at the corresponding time step (including the starting configuration `agents[0]`). An `Agent` object stores all attributes of an agent.
- `positions` is a numpy array of shape `(timesteps+1, n_x, n_y)`, storing for each timestep the number of agents in each grid cell of the domain.
- `food` is a numpy array like `positions` that stores the number of food items per grid cell and time step.

```python
timesteps = 100
agents, positions, food = model.simulate(timesteps, live_plot=True, plot_interval=10)
```

To quickly get the (x, y) positions or (m, d, r) parameter values of an agent population (a list of agents), you can use the functions `get_agent_positions(agents)` and `get_agent_properties(agents)`.

```python
xx, yy = get_agent_positions(agents[0])

import matplotlib.pyplot as plt
plt.scatter(xx, yy)
plt.show()
```
