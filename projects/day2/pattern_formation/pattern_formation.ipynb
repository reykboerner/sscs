First, let's load the source code for the cellular automaton simulator, which is contained in the `CellularAutomaton` and `CA_Parameters` classes in the file `cellular_automaton.py`.

Additionally, we'll need the `numpy` package.

```python
import numpy as np
from cellular_automaton import CellularAutomaton, CA_Parameters
```

The `CA_Parameters` class allows us to specify the parameter settings. You can either use the default values, `CA_Parameters()`, or change the parameters by passing them as keyword arguments. For example, if you want to change the value of `p` to 0.6, you can write

`par = CA_Parameters(p=0.6)`.

```python
par = CA_Parameters()
par.show()
```

Next, we need to define the evolution rule. The following function implements the activation/inhibition rule given in the exercise.

For the code to work, the function must have exactly these arguments in the given order.

```python
def rule(i, j, A, neighbors_a, neighbors_i, parameters):
    """
    Evolution rule of the CellularAutomaton. 

    Arguments
    ---------
        - i: cell x coordinate
        - j: cell y coordinate
        - A: (N, M) lattice with cell states (1 - active, 0 - passive)
        - neighbors_a: list of cells within activator range R_a
        - neighbors_i: list of cells within inhibitor range R_i
        - parameters: instance of CA_Parameters

    Returns either 1 (site is active) or 0 (site is passive)
    """
    w_a = parameters.w_a
    w_i = parameters.w_i

    s = np.sum(A[neighbors_a]*w_a) - np.sum(A[neighbors_i]*w_i)
    return int(s>0) # returns 1 if s>0, else 0
```

Now that we specified the parameters and the evolution rule, we can create an instance of the cellular automaton simulator:

```python
ca = CellularAutomaton(par, rule)
```

To run a simulation, we use the `simulate` function.

```python
N_timesteps = 50
sim = ca.simulate(N_timesteps)
```
