```python
from power_grid import KuramotoPowerGrid
import numpy as np
```

```python
N = 10
K = 0.15

seed = np.random.seed(123)

adjacency_matrix = np.ones((N,N)) - np.eye(N)
power_inputs = np.random.rand(N)*0.2
```

```python
print(power_inputs)
```

```python
model = KuramotoPowerGrid(N, K, adjacency_matrix, power_inputs)
```

```python
angles, time = model.simulate(100.0)
```

```python
import matplotlib.pyplot as plt
```

```python
for i in range(N):
    plt.plot(angles[i])
#plt.xlim(0,5000)
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 5))
gs = fig.add_gridspec(2, 2, width_ratios=[1, 1])

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[1, 0], sharex=ax1)
ax3 = fig.add_subplot(gs[:, 1])

ax1.plot(time, angles[0])
ax2.plot(time, model.order_parameter(angles))

circle = plt.Circle((0, 0), 1, fill=False, color='k')
ax3.add_patch(circle)
ax3.scatter(np.cos(angles[0]), np.sin(angles[0]))

ax3.set_aspect('equal')
ax3.set_xlim(-1.1, 1.1)
ax3.set_ylim(-1.1, 1.1)

plt.tight_layout()
plt.show()
```
