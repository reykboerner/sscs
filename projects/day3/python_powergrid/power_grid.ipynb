# Synchronization in power grids

```python
## DO NOT CHANGE THIS
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from IPython.display import display, HTML
display(HTML("<style>div.output_scroll { height: 44em; }</style>"))
```

Let's load the `KuramotoPowerGrid` class from the file `power_grid.py` (must be in the same folder as this notebook).

```python
from power_grid import KuramotoPowerGrid
```

Next, we set up the power grid.

```python
N = 10      # number of nodes
K = 0.15    # line capacity

# Fix the random number generator seed for reproducible results
seed = np.random.seed(123)

# Construct the adjacency matrix
adjacency_matrix = np.ones((N,N)) - np.eye(N)

# Define the power inputs/outputs
power_usage_level = 0.2
power_inputs = np.random.rand(N)*power_usage_level
```

With these settings, we can build the model. This immediately plots the network structure and power distribution for reference.

```python
model = KuramotoPowerGrid(N, K, adjacency_matrix, power_inputs)
```

Use the `simulate` function to run a simulation. It returns three things: the phase angles of all oscillators over time, the time points, and the order parameter over time.

```python
total_time = 100.0
angles, time, order_param = model.simulate(total_time, live_plot=True)
```

To construct Watts-Strogatz networks, use the code below.

```python
# Create a Watts–Strogatz network
G = nx.watts_strogatz_graph(
    n=100,   # number of nodes
    k=4,     # each node initially connected to 4 neighbors
    p=0.1,   # probability of rewiring
    seed=42
)

# Extract adjacency matrix
A = nx.to_numpy_array(G)

# Draw the network
nx.draw(
    G,
    node_size=30,
    node_color="steelblue",
    edge_color="gray",
    with_labels=False
)

plt.show()
```
